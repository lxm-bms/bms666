```python
#实验2
def score_level(score):
    if not (0 <= score <=100):
        raise ValueError("分数必须0‑100")
    if 90 <= score <=100:
        return "优秀"
    elif 80 <= score <90:
        return "良好"
    elif 60 <= score <80:
        return "及格"
    else:
        return "不及格"

#for求和
s_for = 0
for i in range(1,101):
    s_for +=i
print("for求和1‑100：",s_for)

#while求和
s_while =0
i=1
while i<=100:
    s_while +=i
    i+=1
print("while求和1‑100：",s_while)

#迭代阶乘
def fact_iter(n):
    res=1
    for k in range(2,n+1):
        res *=k
    return res

#递归阶乘
def fact_recur(n):
    if n ==0 or n==1:
        return 1
    return n * fact_recur(n-1)

for n in [0,1,5,10]:
    print(f"n={n},迭代:{fact_iter(n)},递归:{fact_recur(n)}")

def summary(values):
    minv = min(values)
    maxv = max(values)
    avgv = sum(values)/len(values)
    return (minv,maxv,avgv)

res = summary([10,20,30])
mi,ma,av = res #元组解包
print(mi,ma,av)

def sum_all(*args):
    return sum(args)
print(sum_all())
print(sum_all(5))
print(sum_all(1,2,3,4))
```

```python
import os
import pickle

print("当前工作目录", os.getcwd())
data_dir = os.path.join(os.getcwd(),"exp01_data")
os.makedirs(data_dir, exist_ok=True)

txt_path = os.path.join(data_dir,"students.txt")
report_path = os.path.join(data_dir,"report.txt")
pkl_path = os.path.join(data_dir,"students.pkl")

#读取文件处理
valid_records = []
error_count = 0
valid_count =0

from student_utils import parse_student

with open(txt_path,"r",encoding="utf-8") as f:
    lines = f.readlines()
for line_no,line in enumerate(lines, start=1):
    if line_no ==1:
        continue #跳过表头
    try:
        rec = parse_student(line)
        valid_records.append(rec)
        valid_count +=1
    except ValueError as e:
        error_count +=1
        print(f"行{line_no}错误：{e}")

#统计有效记录
scores_valid = [r["score"] for r in valid_records]
avg_score = sum(scores_valid)/len(scores_valid)
pass_num = sum(1 for s in scores_valid if s>=60)
pass_rate = pass_num / len(scores_valid)

# w模式写入report.txt
content = f"""有效记录:{valid_count}
无效记录:{error_count}
平均分:{avg_score:.2f}
及格率:{pass_rate:.2%}
"""
with open(report_path,"w",encoding="utf-8") as fw:
    fw.write(content)

# a追加模式演示
with open(report_path,"a",encoding="utf-8") as fa:
    fa.write("---追加一行测试内容---\n")

#pickle保存
with open(pkl_path,"wb") as fout:
    pickle.dump(valid_records, fout)

#pickle读取校验
with open(pkl_path,"rb") as fin:
    load_back = pickle.load(fin)
print("pickle复原是否相等", load_back == valid_records)

#捕获文件不存在异常
try:
    open("not_exist.txt","r")
except FileNotFoundError as e:
    print("捕获文件不存在异常：",e)

#测试parse_student函数
print(parse_student("Tom,88"))
try:
    parse_student("Tom,200")
except ValueError as e:
    print("无效样例报错：",e)
```

```python
import copy

#1 Student类
class Student:
    def __init__(self,name,score):
        self.name = name
        self.score = score
    def is_passed(self):
        return self.score >=60
    def __str__(self):
        return f"Student(name={self.name},score={self.score})"

#2 根据有效记录创建对象
for r in valid_records:
    s_obj = Student(r["name"],r["score"])
    print(s_obj, "及格" if s_obj.is_passed() else "不及格")

#3 子类GraduateStudent继承
class GraduateStudent(Student):
    def __init__(self,name,score,research_topic):
        super().__init__(name,score)
        self.research_topic = research_topic
    def __str__(self):
        return f"GraduateStudent(name={self.name},score={self.score},topic={self.research_topic})"

g = GraduateStudent("Li",80,"AI")
print(g)

#4深浅拷贝对比
origin = {"name":"Alice","scores":[85,90]}
#直接赋值
obj1 = origin
#浅拷贝
obj2 = copy.copy(origin)
#深拷贝
obj3 = copy.deepcopy(origin)

obj1["scores"][0]=999
print("原始",origin)
print("直接赋值obj1",obj1)
print("copy.copy obj2",obj2)
print("deepcopy obj3",obj3)
```
