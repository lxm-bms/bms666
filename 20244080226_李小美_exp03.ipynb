## 实验一 正则表达式的基础匹配

```python
import re
```

```python
text = "cat on mat; later: cats and dogs"
print(re.findall(r"\d+", text))
print(re.findall(r"\s+", text))
print(re.findall(r"\bcat\b", text))
#说明：\d+ 匹配数字（此文本无数字，结果为空）；\s+ 匹配空白；\bcat\b 只匹配完整单词 cat，不匹配 cats。
```

```python
text = "cat dog bird CAT"
print(re.findall(r"cat|dog", text))
print(re.findall(r"a|aa|aaa", "aaaa"))
print(re.findall(r"aaa|aa|a", "aaaa"))
#说明：交替按分支书写顺序尝试。a|aa|aaa 每次先匹配 a，得到 4 个 a；aaa|aa|a 先匹配最长 aaa，得到 ['aaa', 'a']。
```

```python
text = "ct cat caat caaat"
for p in [r"ca*t", r"ca+t", r"ca?t", r"ca{2}t", r"ca{1,2}t"]:
    print(p, "->", re.findall(p, text))
```

## 实验二 分组，回溯引用与替换

```python
text = "联系: alice@example.com; 备选: bob@test.org"
matches = re.findall(r"([\w.+-]+)@([\w.-]+\.\w+)", text)
print(matches)
for user, domain in matches:
    print(f"用户名: {user}, 域名: {domain}")
```

```python
text = "cat cat, dog cat, bird bird"
print(re.findall(r"\b(\w+)\s+\1\b", text))
```

```python
text = "2024-05-01\n2024/05/02\n2024.05.03"
print("转换前：")
print(text)
converted = re.sub(r"(\d{4})[-/.](\d{2})[-/.](\d{2})", r"\2-\3-\1", text)
print("转换后：")
print(converted)
```

```python
text = "联系我: 123-456-7890 或 987.654.3210"
print(re.sub(r"(\d{3})[-.](\d{3})[-.](\d{4})", r"(\1) \2-\3", text))
```

## 实验三 使用BeautifulSoup 解析 HTML

```python
from bs4 import BeautifulSoup

html_doc = """<html><head><title>实验课程</title></head><body>
<h1>数据分析</h1>
<p class="course">第六讲: 正则表达式</p>
<p class="course">第七讲: 结构化数据</p>
<a href="https://example.org/lecture06">讲义六</a>
<a href="https://example.org/lecture07">讲义七</a>
</body></html>"""

soup = BeautifulSoup(html_doc, "html.parser")
print("标题文本:", soup.title.get_text())
print("h1 标签名:", soup.h1.name)
print("h1 文本:", soup.h1.get_text())
```

```python
courses = soup.find_all("p", class_="course")
for p in courses:
    print("段落:", p.get_text())

links = soup.find_all("a")
for a in links:
    print("链接文字:", a.get_text(), "| href:", a["href"])
```

```python
print("整个文档:", repr(soup.get_text()))
print("单个段落:", repr(courses[0].get_text()))
print("单个链接:", repr(links[0].get_text()))
#说明：整文档 get_text() 返回全部可见文本（含标题、h1、段落、链接文字，保留换行）；单个段落只返回该段文字；单个链接只返回链接文字，不含 href
```
