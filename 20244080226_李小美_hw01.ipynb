```python
def gcd(a,b):
    while b != 0:
        a, b = b, a % b
    return a

print(gcd(2018,2019))
print(gcd(1200,300))
print(gcd(5040,60))
```

```python
import math

def euler_limit(n):
    res = 1
    for i in range(1,n+1):
        res *= i
    return res

def eulaer_infinite_sum(n):
    e = 0
    for k in range(0,n+1):
        e += 1 / euler_limit(k)
    return e

def euler_approx(n):
    return math.e - eulaer_infinite_sum(n)

def print_eular_sum_table(max_n):
    print("n\t估算e\t差值")
    for n in range(1,max_n+1):
        print(f"{n}\t{eulaer_infinite_sum(n):.8f}\t{euler_approx(n):.8f}")

print_eular_sum_table(10)
```

```python
def is_power_of_2(n):
    while n % 2 == 0:
        n = n // 2
    return n == 1

# 测试
print(is_power_of_2(1))  
print(is_power_of_2(8))   
print(is_power_of_2(9))   
print(is_power_of_2(16)) 
```

```python
def is_power(b, n):
    while n % b == 0:
        n = n // b
    return n == 1

print(is_power(2,8))  
print(is_power(3,81)) 
print(is_power(2,10))  
print(is_power(5,1)) 
```
